```python
import pandas as pd
import locale
import datetime as dt
import numpy as np
```

```python
locale.setlocale(locale.LC_TIME, 'fr_FR.UTF-8') 
```

```python
lignes_file = 'correspondances_date_non_formatees.xlsx'

sheets = ['Ligne 1', 'Ligne 2',  'Ligne 2 Sud','Ligne 3', 'Ligne 4', 'Ligne 5', 'Ligne 6', 'Ligne 7', 'Ligne 8', 
          'Ligne 9', 'Ligne 10', 'Ligne 11', 'Ligne 12', 'Ligne 13', 'Ligne 14', 'Ligne 3bis', 'Ligne 7bis', 
          'Ligne 14 (ancienne)', 'Couloirs', 'Voie des Fêtes et voie navette']
```

```python
lignes = []
for sheet in sheets: 
    df = pd.read_excel(lignes_file, sheet_name=sheet)
    df['Ligne'] = sheet
    df['Ouverture'] = df['Ouverture'].map(lambda x: dt.datetime.strptime(x.replace('aout', 'août'), '%d %B %Y'))
    df['Fermeture'] = df['Fermeture'].map(lambda x: dt.datetime.strptime(x, '%d %B %Y') if type(x)==str else x)
    lignes.append(df)
lignes = pd.concat(lignes).reset_index(drop=True)
```
