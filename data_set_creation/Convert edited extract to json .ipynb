```python
import pandas as pd 
```

```python
df = pd.read_csv('station_extract_with_edits.csv')
df.set_index('Station', inplace=True)
```

```python
station_core = df[['latitude', 'longitude']]

correspondance_l = []
correspondance_precedente_c = ['ligne', 'mise en service', 'station précédente']
correspondance_suivante_c = ['ligne', 'mise en service', 'station suivante']
correspondance_c = ['ligne', 'mise en service', 'correspondance']
for i in range(1,6):
    _correspondance_precedente = df[['{} {}'.format(c, str(i)) for c in correspondance_precedente_c]]
    _correspondance_precedente.columns = correspondance_c
    correspondance_l.append(_correspondance_precedente)
    
    _correspondance_suivante = df[['{} {}'.format(c, str(i)) for c in correspondance_suivante_c]]
    _correspondance_suivante.columns = correspondance_c
    correspondance_l.append(_correspondance_suivante)
    
correspondance = pd.concat(correspondance_l).reset_index()
```

```python
correspondance.to_json(orient='records')
```
