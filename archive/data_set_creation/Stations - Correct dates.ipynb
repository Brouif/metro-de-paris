```python
import pandas as pd
import locale
import datetime as dt
import numpy as np
```

```python
locale.setlocale(locale.LC_TIME, 'fr_FR.UTF-8') 
```

```python
def string_to_datetime(x):
    # already formated 
    try: 
        datetime = dt.datetime.strptime(x.replace('0:00:00', ''), '%m/%d/%Y ')
    except: 
        datetime = dt.datetime.strptime(x.replace('1er', '1'), '%d %B %Y')
    
    return datetime
```

```python
station_file = 'station_info_avec_changement_de_nom.csv'
```

```python
stations = pd.read_csv(station_file, index_col=0)

stations['Ouverture'] = stations['Ouverture'].map(lambda x: string_to_datetime(x) if type(x)==str else x)
stations['Fermetrue'] = stations['Fermetrue'].map(lambda x: string_to_datetime(x) if type(x)==str else x)
```

```python
stations.to_csv('station_info_avec_changement_de_nom_date_formatted.csv')
```
