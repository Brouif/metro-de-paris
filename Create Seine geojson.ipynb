```python
import geopandas 
import fiona 
```

```python
geopandas.io.file.fiona.drvsupport.supported_drivers['KML'] = 'rw'
```

```python
file_path = 'data/raw_data/'
file_name = 'PLAN_EAU'
```

```python
shapes = geopandas.read_file(file_path+file_name+'.kml')
info = geopandas.read_file(file_path+file_name+'.csv').drop('geometry', axis=1)

df = info.join(shapes)
```

```python
libele_eau = 'Seine'
```

```python
seine = df[df['L_EAU']=='Seine']
```

```python
output_path = 'data/'
output_name = 'seine.geojson'
```

```python
seine.to_file(output_path+output_name, driver='GeoJSON')
```
